#### API-TEST-Multiple parameter
An API Request for the Wien Hohe Warte with multiple Parameters
TL - Temperature
FFAM - Arithmetisches Mittel der Windgeschwindigkeit
RF - Relative Feuchte

```python
import requests
import json

url = "https://dataset.api.hub.geosphere.at/v1/station/current/tawes-v1-10min"
params = {
    "parameters": "TL,FFAM,RF",
    "station_ids": "11035"
}
response = requests.get(url, params=params)
print(json.dumps(response.json(), indent = 2))
```

### Get List of all Station_IDs and City names
# for the state "Kärnten"
Is for Human reading to make a decision, which stations should be picked for a project.

```python
import json


with open('metadaten_hourly.json') as df:
    file = json.load(df)

    with open("station_infos.txt", "w", encoding= 'utf-8') as text_file:
        for station in file["stations"]:
            if station["state"] == "Kärnten":
                data = {
                    "name" : station["name"],
                    "state": station["state"],
                    "station_id": station["id"]
                }
            
                text_file.write(json.dumps(data, ensure_ascii=False) + "\n")        
            else:
                continue
```
